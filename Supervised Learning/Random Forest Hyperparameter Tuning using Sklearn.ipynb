```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
```

```python
data = pd.read_csv('Data/heart.csv')
```

```python
data.head(7)
```

```python
X = data.drop('target', axis=1)
y = data['target']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.25, random_state=42)
model = RandomForestClassifier(n_estimators=100, max_features = "sqrt", max_depth=6,
                              max_leaf_nodes=6)
```

```python
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
print(classification_report(y_pred, y_test))
```

**1. Hyperparameter Tuning using GridSearchCV**

```python
param_grids = {'n_estimators':[100, 200],
                'max_depth':[None, 10, 20],
                'min_samples_split':[2,5],
                'min_samples_leaf':[1,2],
                'bootstrap':[True, False]
}
grid_search = GridSearchCV(RandomForestClassifier(), param_grid = param_grids, cv=5)
grid_search.fit(X_train, y_train)
```

```python
print("Best Parameters:", grid_search.best_params_)
print("Best Estimators:", grid_search.best_estimator_)
```

```python
model_grid = RandomForestClassifier(max_depth=3,
                                   max_features='log2',
                                   max_leaf_nodes = 3,
                                   n_estimators=50)
model_grid.fit(X_train, y_train)
y_pred_grid = model.predict(X_test)
print(classification_report(y_pred_grid, y_test))
```

**2. Hyperparameter Tuning using RandomizedSearchCV**

```python
random_search = RandomizedSearchCV(RandomForestClassifier(), param_grids)
random_search.fit(X_train, y_train)
print(random_search.best_estimator_)
```

```python
model_random = RandomForestClassifier(max_depth=3,
                                     max_features='log2',
                                     max_leaf_nodes=6,
                                     n_estimators=100)
model_random.fit(X_train, y_train)
y_pred_rand = model.predict(X_test)
print(classification_report(y_pred_rand, y_test))
```
