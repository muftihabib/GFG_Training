**1: Importing necessary Libraries**

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from matplotlib import gridspec
from sklearn.model_selection import train_test_split
```

**2: Loading the Data**

```python
data = pd.read_csv("Data/creditcard.csv")
print(data.head())
```

```python
data.describe()
```

**3: Analyzing Class Distribution**

```python
fraud = data[data['Class']==1]
valid = data[data['Class']==0]
outlierFraction = len(fraud)/float(len(valid))
print(outlierFraction)
print('Fraud Cases:{}'.format(len(data[data['Class']==1])))
print('Valid Transaction:{}'.format(len(data[data['Class']==0])))
```

**4: Exploring Transaction Amounts**

```python
print("Amount details of the fraudlent transaction")
fraud.Amount.describe()
```

```python
print("details of valid transaction")
valid.Amount.describe()
```

**5: Plotting Correlation Matrix**

```python
corrmat = data.corr()
fig = plt.figure(figsize=(12,9))
sns.heatmap(corrmat, vmax=.8, square=True)
plt.show()
```

**6: Preparing Data**

```python
X = data.drop(['Class'], axis=1)
y = data['Class']
print(X.shape)
print(y.shape)

xData = X.values
yData = y.values
```

```python
X_train, X_test, y_train, y_test = train_test_split(xData, yData, test_size=0.2, random_state=42)
```

**7: Building and Training the Model**

```python
from sklearn.ensemble import RandomForestClassifier
rfc = RandomForestClassifier()
rfc.fit(X_train, y_train)
y_pred = rfc.predict(X_test)
```

**8: Evaluating the Model**

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score,matthews_corrcoef, confusion_matrix
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
mcc = matthews_corrcoef(y_test, y_pred)

print("Model Evaluation Metrics:")
print(f"Accuarcy:{accuracy:.4f}")
print(f"Precision:{precision:.4f}")
print(f"Recall:{recall:.4f}")
print(f"F1-Score:{f1:.4f}")
print(f"Matthews Correlation Coefficient:{mcc:.4f}")

conf_matrix = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(8,6))
sns.heatmap(conf_matrix, annot=True,fmt='d',cmap="Blues", xticklabels=['Normal','Fraud'],
           yticklabels=['Normal','Fraud'])
plt.title("Confusion Matrix")
plt.xlabel("Predicted Class")
plt.ylabel("True Class")
plt.show()
```
